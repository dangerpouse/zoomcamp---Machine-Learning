```python
xi = [453, 11, 86]
w0 = 7.17#average price(logged). Think this acts as the Y intercept
w = [0.01, 0.04, 0.002]
```

```python
def linear_regression(xi):
    n = len(xi)

    pred = w0

    for j in range(n):
        pred = pred + w[j] * xi[j]

    return pred
```

Valuation has been logged during previous session. Use expm1 to unlog

```python
pred = linear_regression(xi)
pred_final =  np.expm1(pred)
pred_final
```
