```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data_prepped.csv')
```

```python
df.head()
```

Get splits for training, validation and testing

```python
n = len(df)
n_val = int(n*0.2)
n_test = int(n*0.2)
n_train = n - n_val - n_test
print(f"Total Records:{n}, Validation Records:{n_val}, Testing Records:{n_test}, Training Records:{n_train}, Total:{n_val+n_test+n_train}")
```

Need to shuffle the records since they are ordered by make

```python
idx = np.arange(n)
np.random.seed(2)
np.random.shuffle(idx)
```

Now get records based on shuffled index and then reset index to make the index readible

```python
df_val = df.iloc[idx[:n_val]]
df_test = df.iloc[idx[:n_test]]
df_train = df.iloc[idx[:n_train]]

len(df_val), len(df_test), len(df_train)

df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
df_train = df_train.reset_index(drop=True)
```

```python
del df_val['Unnamed: 0']
del df_test['Unnamed: 0']
del df_train['Unnamed: 0']
```

Put the target variable(msrp) in a separate vector and then delete it because we do not want the model to be trained with it

```python
y_val = np.log1p(df_val['msrp'].values)
y_test = np.log1p(df_test['msrp'].values)
y_train = np.log1p(df_train['msrp'].values)
```

```python
del df_val['msrp']
del df_test['msrp']
del df_train['msrp']
```

```python
df_val.to_csv('data_validation.csv', index=False)
df_test.to_csv('data_test.csv', index=False)
df_train.to_csv('data_train.csv', index=False)
```
