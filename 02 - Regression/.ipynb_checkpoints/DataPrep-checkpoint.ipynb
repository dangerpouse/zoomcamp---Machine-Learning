```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('data.csv')
df.head(5)
```

```python
df.columns=df.columns.str.lower().str.replace(' ','_')
```

```python
df.isnull().sum()
```

```python
sns.countplot(df,y='engine_fuel_type',hue='engine_fuel_type', legend=False)
```

```python
strings = list(df.dtypes[df.dtypes=='str'].index)
df.dtypes[df.dtypes=='str'].index
```

```python
df[strings] = df[strings].apply(lambda col: col.str.lower().str.replace(' ','_'))
```

```python
df.to_csv('data_prepped.csv')
```
