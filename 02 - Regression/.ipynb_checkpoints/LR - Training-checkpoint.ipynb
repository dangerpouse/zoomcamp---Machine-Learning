```python
import pandas as pd
import numpy as np
```

```python
df_train = pd.read_csv('data_train.csv')
df_test = pd.read_csv('data_test.csv')
df_val = pd.read_csv('data_validation.csv')
```

```python
#generate feature matrix
X = [
    [148, 24, 1385],
    [132, 25, 2031],
    [453, 11, 86],
    [158, 24, 185],
    [172, 25, 201],
    [413, 11, 86],
    [38,  54, 185],
    [142, 25, 431],
    [453, 31, 86],
]
X = np.array(X)
```

```python
XTX = X.T.dot(X)#This will get the Gram matrix. Results will be a square which can then be used to get the inverse
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
#Can check if this has worked by seeing if we can get the identity matrix
XTX.dot(XTX_inv).round(1)#Use round(1) because numbers are very small so may not appear to be an indetity matrix
```

Generate sample target vector. Observed prices for our feature matrix

```python
y = [100, 200, 150, 250, 100, 200, 150, 250, 120]
```

```python
w_full = XTX_inv.dot(X.T).dot(y)#These are ouer weights. They are missing the Y intercept though
w_full
```

```python
#Lets add 1's to the feature matrix
ones = np.ones(X.shape[0])
X = np.column_stack([ones,X])
X
```

```python
#Now redo the weights
#w_full = (XtX)-1.(X.t).y
XTX = X.T.dot(X)
XTX_inv = np.linalg.inv(XTX)
XTX.dot(XTX_inv).round(1)
w_full = XTX_inv.dot(X.T).dot(y)
w_full
```

```python
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)

    return w_full[0], w_full[1:]
```

Get the weights and bias term(Y intercept)

```python
train_linear_regression(X, y)
```
